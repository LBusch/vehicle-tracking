```python
# @title Extract data into colab environment
'''
  make sure to zip the DETRAC data folder and upload it to your google drive
'''

import shutil

ZIP_PATH = "/content/drive/MyDrive/DETRAC.zip"  # path of zip file on drive
TARGET_PATH = "/content/DETRAC"                 # target path in colab

print("Extracting...")
shutil.unpack_archive(ZIP_PATH, TARGET_PATH)
print(f"Successfully extracted to: {TARGET_PATH}")
```

```python
# @title Train YOLO model

from ultralytics import YOLO
import torch
import os

assert torch.cuda.is_available(), "No GPU available!"

'''
  make sure to select project path on drive
  to save results and checkpoints on drive
'''
PROJECT_PATH = "/content/drive/MyDrive/vehicle_detection"
DATA_YAML = "/content/DETRAC/DETRAC.yaml"
RUN_NAME = "yolo26s"
N_EPOCHS = 5
BATCH_SIZE = -1   # auto batch
IMAGE_SIZE = 640

RESUME = False    # resume interrupted training
WEIGHT_DIR = os.path.join(PROJECT_PATH, RUN_NAME, "weights", "last.pt")

if RESUME:
  model = YOLO(WEIGHT_DIR)    # saved YOLO weights
else:
  model = YOLO("yolo26s.pt")  # pretrained YOLO weights

results = model.train(
    data=DATA_YAML,
    epochs=N_EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    device=0,
    workers=8,
    project=PROJECT_PATH,
    name=RUN_NAME,
    exist_ok=True,
    resume=RESUME,
)
```
